```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
def bell() -> QuantumCircuit:
    qc = QuantumCircuit(2, name="bell")
    qc.h(0)
    qc.cx(0, 1)

    return qc

prep = bell()
```

```python
prep.draw("mpl", style="iqp")
```

```python
Statevector(prep).draw("qsphere")          # amplitudes and phases on one sphere
```

```python
def alice_encode(msg: int) -> QuantumCircuit:
    qc = QuantumCircuit(1, name=f"Alice {msg:02b}")
    match msg:
        case 0b00: qc.id(0)
        case 0b01: qc.x(0)
        case 0b10: qc.z(0)
        case 0b11: qc.x(0); qc.z(0)
        case _: raise ValueError(msg)
    return qc

def bob_decode() -> QuantumCircuit:
    qc = QuantumCircuit(2, name="bob")
    qc.cx(0, 1)
    qc.h(0)
    
    return qc
```

```python
def protocol(message: int) -> QuantumCircuit:
    qc = QuantumCircuit(2, name="protocol")
    qc.compose(bell(), inplace=True)
    qc.barrier()
    qc.append(alice_encode(message).to_gate(label=f"alice"), [0])
    qc.barrier()
    qc.compose(bob_decode(), inplace=True)
    return qc
```

```python
for message in [0b00, 0b01, 0b10, 0b11]:
    probs = Statevector(protocol(message)).probabilities_dict()
    probs = {k[::-1]: round(v, 6) for k, v in probs.items()}
    print(f"{message:02b} ->", probs)
```
